## Zadanie 2

W jaki sposób należy wykonać normalizację takich danych? **Napisz funkcję**, która na to pozwoli.


```python
import numpy as np

def norm(data):
    minn = np.min(data, axis=0)
    maxx = np.max(data, axis=0)
    return (data - minn) / (maxx - minn)
print(norm(data))
```

**Uzupełnij:** 
- `Dense` służy do tworzenia warstwy w pełni połączonej (każdy neuron łączy się ze wszystkimi wyjściami poprzedniej warstwy). Jest to warstwa, która będzie uczona, jej wagi i biasy będą się zmieniać.
- `Activation` służy do nałożenia funkcji aktywacji (np. ReLU), która wprowadza nieliniowość. Jest to warstwa, która nie będzie uczona, jej wartości nie będą się zmieniać.
- `Dropout` służy do regularyzacji: w trakcie treningu losowo wyłącza część neuronów, co zmniejsza przeuczenie. Jest to warstwa, która nie będzie uczona.
- `Sequential` służy do budowy modelu, w którym warstwy układamy jedna po drugiej.
- `to_categorical` służy do zamiany etykiet na kodowanie one-hot.
